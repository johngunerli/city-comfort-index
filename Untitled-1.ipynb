```python
import pandas as pd
import numpy as np
import os
import plotly.express as px

# === CONFIG ===
DATA_DIR = "data"
CITY_INFO_FILE = os.path.join(DATA_DIR, "city_info.csv")

START_YEAR = 2000
END_YEAR = 2019
LOW, HIGH = 65, 75  # °F comfort range
MIDPOINT = 70  # midpoint for comfort scoring

CITY_STATE_MAP = {
    "Lander": "WY",
    "Cheyenne": "WY",
    "Wausau": "WI",
    "Hayward": "WI",
    "EauClaire": "WI",
    "Ashland": "WI",
    "Milwaukee": "WI",
    "Madison": "WI",
    "LaCrosse": "WI",
    "GreenBay": "WI",
    "Bluefield": "WV",
    "Morgantown": "WV",
    "Wheeling": "WV",
    "Huntington": "WV",
    "Charleston": "WV",
    "Beckley": "WV",
    "Vancouver": "WA",
    "Pullman": "WA",
    "TriCities": "WA",
    "Spokane": "WA",
    "Seattle": "WA",
    "Richmond": "VA",
    "NorfolkVA": "VA",
    "Lynchburg": "VA",
    "Burlington": "VT",
    "SaltLakeCity": "UT",
    "CollegeStation": "TX",
    "SanAntonio": "TX",
    "Houston": "TX",
    "Galveston": "TX",
    "ElPaso": "TX",
    "DallasFortWorth": "TX",
    "CorpusChristi": "TX",
    "Austin": "TX",
    "Amarillo": "TX",
    "Abilene": "TX",
    "Nashville": "TN",
    "Memphis": "TN",
    "Chattanooga": "TN",
    "Mitchell": "SD",
    "SiouxFalls": "SD",
    "Huron": "SD",
    "Aberdeen": "SD",
    "Aiken": "SC",
    "GreenvilleSpartanburg": "SC",
    "ColumbiaSC": "SC",
    "Williamsport": "PA",
    "Pittsburgh": "PA",
    "Philadelphia": "PA",
    "MiddletownHarrisburg": "PA",
    "Erie": "PA",
    "KlamathFalls": "OR",
    "Portland": "OR",
    "Salem": "OR",
    "Pendleton": "OR",
    "EUGENE": "OR",
    "Astoria": "OR",
    "Stillwater": "OK",
    "Tulsa": "OK",
    "Oklahoma": "OK",
    "Zaneville": "OH",
    "Toledo": "OH",
    "Dayton": "OH",
    "ColumbusOH": "OH",
    "Cleveland": "OH",
    "Cincinnati": "OH",
    "Akron": "OH",
    "Williston": "ND",
    "GrandForks": "ND",
    "Fargo": "ND",
    "Bismarck": "ND",
    "Raleigh": "NC",
    "Charlotte": "NC",
    "CapeHatteras": "NC",
    "Asheville": "NC",
    "GlensFalls": "NY",
    "RochesterNY": "NY",
    "NewYork": "NY",
    "Buffalo": "NY",
    "Albany": "NY",
    "Roswell": "NM",
    "Clayton": "NM",
    "Albuquerque": "NM",
    "Belvidere": "NJ",
    "NewBrunswick": "NJ",
    "Trenton": "NJ",
    "Manchester": "NH",
    "Concord": "NH",
    "Winnemucca": "NV",
    "Reno": "NV",
    "Ely": "NV",
    "Elko": "NV",
    "McCook": "NE",
    "Hastings": "NE",
    "Valentine": "NE",
    "Scottsbluff": "NE",
    "Omaha": "NE",
    "NorfolkNE": "NE",
    "NorthPlatte": "NE",
    "Lincoln": "NE",
    "GrandIsland": "NE",
    "Missoula": "MT",
    "Helena": "MT",
    "Havre": "MT",
    "GreatFalls": "MT",
    "Glasgow": "MT",
    "Jefferson": "MO",
    "StLouis": "MO",
    "SpringfieldMO": "MO",
    "KansasCity": "MO",
    "ColumbiaMO": "MO",
    "Gulfport": "MS",
    "Hattiesburg": "MS",
    "Meridian": "MS",
    "Jackson": "MS",
    "SaintCloud": "MN",
    "RochesterMN": "MN",
    "Minneapolis": "MN",
    "InternationalFalls": "MN",
    "Duluth": "MN",
    "Kalamazoo": "MI",
    "Saginaw": "MI",
    "SaultSteMarie": "MI",
    "Muskegon": "MI",
    "Marquette": "MI",
    "Lansing": "MI",
    "GrandRapids": "MI",
    "Detroit": "MI",
    "Worcester": "MA",
    "Milton": "MA",
    "Boston": "MA",
    "Baltimore": "MD",
    "PortlandME": "ME",
    "Natchitoches": "LA",
    "Monroe": "LA",
    "Shreveport": "LA",
    "LakeCharles": "LA",
    "BatonRouge": "LA",
    "Louisville": "KY",
    "Lexington": "KY",
    "Topeka": "KS",
    "Goodland": "KS",
    "DodgeCity": "KS",
    "Concordia": "KS",
    "CedarRapids": "IA",
    "Spencer": "IA",
    "Waterloo": "IA",
    "SiouxCity": "IA",
    "Dubuque": "IA",
    "DesMoines": "IA",
    "SouthBend": "IN",
    "Indianapolis": "IN",
    "FortWayne": "IN",
    "Evansville": "IN",
    "Springfield": "IL",
    "Rockford": "IL",
    "Peoria": "IL",
    "Moline": "IL",
    "Chicago": "IL",
    "Lewiston": "ID",
    "Boise": "ID",
    "Savannah": "GA",
    "Macon": "GA",
    "ColumbusGA": "GA",
    "Augusta": "GA",
    "Atlanta": "GA",
    "Athens": "GA",
    "WestPalmBeach": "FL",
    "Tampa": "FL",
    "Tallahassee": "FL",
    "Pensacola": "FL",
    "Orlando": "FL",
    "Miami": "FL",
    "KeyWest": "FL",
    "Jacksonville": "FL",
    "Gainesville": "FL",
    "FortMyers": "FL",
    "WashingtonDC": "DC",
    "Wilmington": "DE",
    "Pueblo": "CO",
    "GrandJunction": "CO",
    "Denver": "CO",
    "ColoradoSprings": "CO",
    "CrescentCity": "CA",
    "Escondido": "CA",
    "Ukiah": "CA",
    "Sacramento": "CA",
    "Riverside": "CA",
    "Needles": "CA",
    "SanDiego": "CA",
    "Redding": "CA",
    "LosAngeles": "CA",
    "Fresno": "CA",
    "Eureka": "CA",
    "Bakersfield": "CA",
    "Russellville": "AR",
    "PineBluff": "AR",
    "Jonesboro": "AR",
    "Harrison": "AR",
    "LittleRock": "AR",
    "FortSmith": "AR",
    "Yuma": "AZ",
    "Tucson": "AZ",
    "Phoenix": "AZ",
    "Flagstaff": "AZ",
    "MuscleShoals": "AL",
    "Montgomery": "AL",
    "Mobile": "AL",
    "Huntsville": "AL",
    "Birmingham": "AL",
}


# === FUNCTIONS ===
def load_city_metadata():
    info = pd.read_csv(CITY_INFO_FILE)
    info = info.drop_duplicates(subset=["Name", "ID"])
    info["State"] = info["Name"].map(CITY_STATE_MAP)
    if info["State"].isna().any():
        missing = info.loc[info["State"].isna(), "Name"].unique().tolist()
        raise ValueError(f"Missing state mapping for: {missing}")
    return info[["Name", "ID", "Lat", "Lon", "State"]]


def load_city_data(city_id):
    path = os.path.join(DATA_DIR, f"{city_id}.csv")
    if not os.path.exists(path):
        return None
    df = pd.read_csv(path, parse_dates=["Date"])
    df = df[(df["Date"].dt.year >= START_YEAR) & (df["Date"].dt.year <= END_YEAR)]
    if df.empty:
        return None
    df["tavg"] = df["tmax"]  # test: use high temps only
    return df


def average_comfortable_days_per_year(df):
    """Compute the average number of 65–75°F days per year for this city."""
    df = df.dropna(subset=["tavg"])
    df["year"] = df["Date"].dt.year
    yearly_counts = df.groupby("year")["tavg"].apply(
        lambda s: s.between(LOW, HIGH).sum()
    )
    return yearly_counts.mean(), len(yearly_counts)


def compute_comfort_index(df):
    """
    Asymmetric comfort scoring:
      - Ideal: 65–75°F → high score
      - Cooler: mild penalty
      - Hotter: strong penalty
    """
    df = df.dropna(subset=["tavg"]).copy()

    # Asymmetric Gaussian weighting:
    # cooler side (<=70°F) -> sigma=7  | warmer side (>70°F) -> sigma=4
    def comfort_weight(t):
        if t <= MIDPOINT:
            return np.exp(-0.5 * ((t - MIDPOINT) / 7) ** 2)
        else:
            return np.exp(-0.5 * ((t - MIDPOINT) / 2) ** 2)

    df["comfort_score"] = df["tavg"].apply(comfort_weight)

    avg_comfort = df["comfort_score"].mean() * 100  # percent-like scale
    temp_std = df["tavg"].std()
    mean_dev = np.abs(df["tavg"] - MIDPOINT).mean()

    # combine: high comfort, low variability, small deviation = good
    # feel free to tune these weights
    comfort_index = avg_comfort - (0.5 * temp_std) - (0.3 * mean_dev)
    return comfort_index, avg_comfort, temp_std, mean_dev


def analyze_all_cities():
    info = load_city_metadata()
    results = []

    for _, row in info.iterrows():
        city, cid, lat, lon, state = (
            row["Name"],
            row["ID"],
            row["Lat"],
            row["Lon"],
            row["State"],
        )
        df = load_city_data(cid)
        if df is None:
            continue

        avg_comfy, n_years = average_comfortable_days_per_year(df)
        comfort_index, avg_comfort, temp_std, mean_dev = compute_comfort_index(df)

        results.append(
            {
                "city": city,
                "state": state,
                "id": cid,
                "latitude": lat,
                "longitude": lon,
                "avg_comfortable_days_per_year": avg_comfy,
                "years_with_data": n_years,
                "avg_comfort_score_pct": avg_comfort,
                "temp_std_dev": temp_std,
                "mean_deviation_from_70": mean_dev,
                "comfort_index": comfort_index,
            }
        )

    results_df = pd.DataFrame(results)
    results_df = results_df.sort_values(by="comfort_index", ascending=False)
    return results_df


def plot_top_cities(df, top_n=20):
    top = df.head(top_n).copy()
    top["city_label"] = top["city"] + ", " + top["state"]
    fig = px.bar(
        top,
        x="comfort_index",
        y="city_label",
        orientation="h",
        color="comfort_index",
        color_continuous_scale="Viridis",
        title=f"Top {top_n} U.S. Cities by Comfort Index (2000–2019)",
        labels={
            "comfort_index": "Comfort Index (higher = better)",
            "city_label": "City, State",
        },
    )
    fig.update_layout(yaxis={"categoryorder": "total ascending"})
    fig.write_image("top_comfortable_cities.png", scale=2)
    fig.show()


def plot_satellite_map(df):
    df = df.copy()
    df["city_label"] = df["city"] + ", " + df["state"]

    # Normalize comfort_index for marker size (positive range)
    min_ci = df["comfort_index"].min()
    max_ci = df["comfort_index"].max()
    df["size_scaled"] = 5 + 20 * (df["comfort_index"] - min_ci) / (max_ci - min_ci)
    fig = px.scatter_map(
        df,
        lat="latitude",
        lon="longitude",
        color="comfort_index",
        hover_name="city_label",
        size="size_scaled",
        color_continuous_scale="Viridis",
        zoom=3,
        center={"lat": 39.8283, "lon": -98.5795},
        title="U.S. Cities Comfort Index (2000–2019)",
    )

    fig.update_traces(
        text=df["city_label"],
        mode="markers+text",
        textposition="top center",
        textfont={"size": 10, "color": "black"},
    )

    fig.update_layout(
        mapbox_style="satellite-streets",
        mapbox_zoom=3,
        mapbox_center={"lat": 39.5, "lon": -98.35},
        margin={"r": 0, "t": 30, "l": 0, "b": 0},
    )
    fig.write_image("comfortable_cities_map.png", scale=2)
    fig.show()


results = analyze_all_cities()
results.to_csv("comfortable_days_summary.csv", index=False)
print("Saved summary to comfortable_days_summary.csv")

plot_top_cities(results)
plot_satellite_map(results)
```

```python
def plot_state_comfort_map(df):
    """Plot the average comfort index for each U.S. state as a choropleth map."""
    # Compute mean comfort index per state
    state_avg = (
        df.groupby("state", as_index=False)["comfort_index"]
        .mean()
        .rename(columns={"state": "State", "comfort_index": "avg_comfort_index"})
    )

    fig = px.choropleth(
        state_avg,
        locations="State",
        locationmode="USA-states",
        color="avg_comfort_index",
        color_continuous_scale="Viridis",
        scope="usa",
        title="Average Comfort Index by State (2000–2019)",
        labels={"avg_comfort_index": "Average Comfort Index"},
    )

    fig.update_layout(
        geo=dict(bgcolor="rgba(0,0,0,0)"),
        margin={"r": 0, "t": 40, "l": 0, "b": 0},
    )
    fig.write_image("state_comfort_index_map.png", scale=2)
    fig.show()
```
